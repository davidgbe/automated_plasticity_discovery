```python
import numpy as np
import matplotlib
import matplotlib.pyplot as plt
```

$$ u = t_{\rm{post}} - t_z $$
$$ v = t_{\rm{pre}}  - t_z $$
$$ u - v = t_{\rm{post}} - t_{\rm{pre}} $$

Potentiation is highest when postsynaptic activity follows the input *and* postsynaptic activity lags presynaptic activity. These conditions are equivalent to:

$$ u > 0 $$
$$ u - v > 0 $$

A representative triplet potentiating function is:

$$ f(u, v) = e^{-u / \tau_{1}} \, e^{-(u - v) / \tau_{2}} \, \Theta(u) \Theta(u - v) - \frac{1}{4} e^{-|u|/\tau_1} e^{-|v|/\tau_2}$$

When plotted, this looks like:

```python
# parameters
tau1 = 0.01
tau2 = 0.01

# grid
u = np.linspace(-0.02, 0.02, 400)
y = np.linspace(-0.02, 0.02, 400)

U, Y = np.meshgrid(u, y)

# Heaviside
Theta = lambda x: (x >= 0).astype(float)

# function
F = (
    np.exp(-U / tau1)
    * np.exp(-Y / tau2)
    * Theta(U)
    * Theta(Y)
     - 1/4 * np.exp(-np.abs(U) / tau1) * np.exp(-np.abs(Y) / tau2)
)

# plot
plt.figure(figsize=(2, 2))
plt.imshow(
    F,
    extent=[u.min(), u.max(), v.min(), v.max()],
    origin="lower",
    aspect=1,
    cmap='bwr',
    vmax=1,
    vmin=-1,
)
plt.xlabel(r"$u$")
plt.ylabel(r"$y$")
plt.colorbar(label=r"$f\,(u, y)$", pad=0.1, fraction=0.04)
plt.show()
```
